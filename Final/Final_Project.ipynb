## MP4 — Analiz ve Görselleştirme
Algoritma performansı (toplam süre) ile bant genişliği (ciphertext boyutu) arasındaki takas incelenir.

```
import matplotlib.pyplot as plt

numeric_cols = data.select_dtypes(include="number").columns
data[numeric_cols].describe()
```

```
group = (data.groupby(["algorithm","security_level"]).agg(mean_total_time_ms=("total_time_ms","mean"), mean_ciphertext_len=("ciphertext_length","mean"), n=("total_time_ms","size")).reset_index())

group.sort_values("mean_total_time_ms").head(10)
```

```
top_lvl1 = group[group.security_level==1].sort_values("mean_total_time_ms").head(8)
top_lvl1
```

```
plt.figure(figsize=(8,5))
plt.scatter(data["ciphertext_length"], data["total_time_ms"], s=5)
plt.xlabel("Ciphertext Length (bytes)")
plt.ylabel("Total Time (ms)")
plt.title("PQC KEM Benchmark: Süre vs Şifreli Metin Boyutu")
plt.tight_layout()
plt.savefig("kem_performance_regenerated.png", dpi=200)
plt.show()
```
